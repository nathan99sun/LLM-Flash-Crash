```python
import numpy as np
import random

mid_price = 100 # placeholder
```

```python
class NoiseTrader:
    """Traditional stochastic agent to provide baseline volume."""
    def __init__(self, agent_id, arrival_rate=10):
        self.agent_id = agent_id
        self.arrival_rate = arrival_rate # Poisson lambda, adjust as needed

    def step(self, mid_price):
        side = 'buy' if random.random() < 0.5 else 'sell'
        qty = random.randint(10, 100) # adjust quantity as needed
        return {
            "agent_id": self.agent_id,
            "type": "MARKET",
            "side": side,
            "qty": qty
        }
```

Setup (to be included in the simulation)

```python
noise_traders = {f"noise_{i}": NoiseTrader(agent_id=f"noise_{i}") for i in range(50)} # 50 noise traders

# Simulation increment
num_arrivals = min(np.random.poisson(10), len(noise_traders))
selected_traders = random.sample(list(noise_traders.values()), num_arrivals)
orders = [trader.step(mid_price) for trader in selected_traders]
```
